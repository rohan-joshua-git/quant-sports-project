```python
import os
import sys
sys.path.append("../..")

import cv2
from ultralytics import YOLO
from pipeline.common.drawing import annotate_frame

%load_ext autoreload
%autoreload 2

model = YOLO("football_yolo26n_best.pt")
src = r"C:\Users\rohan\Desktop\Quant Sports Project\Tester video\08fd33_4.mp4"

cap = cv2.VideoCapture(src)
fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

os.makedirs("runs", exist_ok=True)
out = cv2.VideoWriter("runs/styled_output.avi", cv2.VideoWriter_fourcc(*"XVID"), fps, (w, h))

while True:
    ok, frame = cap.read()
    if not ok:
        break
    result = model.predict(frame, verbose=False)[0]
    out.write(annotate_frame(frame, result))

cap.release()
out.release()
```
